## 3. Evaluation

This section details the metrics used for evaluating the model's performance and important correctness checks to ensure the robustness of the implementation.

### 3.1 Metrics

**Training Metric:**
*   The cost $J(\theta)$ recorded during gradient descent.

**Test Metric:**
*   Root-Mean-Square Error (RMSE), which is the cost rescaled into kW to compare with the plant's actual power.
*   RMSE is calculated as: $RMSE = \sqrt{\frac{1}{m} \sum_{i=1}^{m} (y^{(i)} - h_{\theta}(x^{(i)}))^2} = \sqrt{2 J_{test}(\theta)}$.
*   Report RMSE for all test hours and for daytime test hours only.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Custom RMSE function built strictly with NumPy (scikit-learn is forbidden if this is from Lecture 1)
def calculate_rmse(y_true, y_pred):
    return np.sqrt(np.mean((y_true - y_pred) ** 2))

try:
    # 1. Load Raw CSVs
    df_gen = pd.read_csv('/content/Plant_1_Generation_Data.csv')
    df_sensor = pd.read_csv('/content/Plant_1_Weather_Sensor_Data (1).csv')

    # 2. Convert timestamps explicitly
    df_gen['datetime'] = pd.to_datetime(df_gen['DATE_TIME'], format='%d-%m-%Y %H:%M')
    df_sensor['datetime'] = pd.to_datetime(df_sensor['DATE_TIME'], format='%Y-%m-%d %H:%M:%S')

    # 3. Aggregate plant-level generation across all inverters (selecting only numeric columns)
    plant_power = df_gen.groupby('datetime')[['AC_POWER', 'DC_POWER']].sum().reset_index()

    # 4. Select relevant numeric columns from sensor data before merging
    sensor_numeric_cols = ['datetime', 'AMBIENT_TEMPERATURE', 'MODULE_TEMPERATURE', 'IRRADIATION']
    df_sensor_numeric = df_sensor[sensor_numeric_cols]

    # 5. Merge plant-level power with cleaned sensor readings
    df_merged = pd.merge(plant_power, df_sensor_numeric, on='datetime', how='inner')

    # 6. Set datetime as index for resampling
    df_merged = df_merged.set_index('datetime')

    # 7. Resample to 1-hour means and handle missing values
    # Explicitly select numeric columns for resampling to avoid 'agg function failed' error
    numeric_cols_for_resample = ['AC_POWER', 'DC_POWER', 'AMBIENT_TEMPERATURE', 'MODULE_TEMPERATURE', 'IRRADIATION']
    df_hourly = df_merged[numeric_cols_for_resample].resample('1h').mean().reset_index()
    df_hourly = df_hourly.interpolate(method='linear').bfill() # Handle missing values after resampling

    # 8. Rename columns to assignment standard for consistency
    df_hourly = df_hourly.rename(columns={
        'AC_POWER': 'ac_power',
        'DC_POWER': 'dc_power',
        'AMBIENT_TEMPERATURE': 'ambient_temp',
        'MODULE_TEMPERATURE': 'module_temp',
        'IRRADIATION': 'irradiation'
    })

    # 9. Chronological Test Split (11 June to 17 June as specified in Task 4 or similar context)
    test_df = df_hourly[df_hourly['datetime'] >= '2020-06-11 00:00:00'].copy()

    # Target AC Power (using renamed column)
    y_true_all = test_df['ac_power'].values

    # --- PLACEHOLDER: Replace this block with your actual model hypothesis (predictions) ---
    # The dummy predictions here simulate model output for demonstration.
    np.random.seed(42) # for reproducibility
    dummy_noise = np.random.normal(0, 50, len(y_true_all)) # Larger noise for demonstration, adjust as needed
    y_pred_all = y_true_all + dummy_noise
    # -------------------------------------------------------------------------------------

    # 10. Clip negative predictions to zero (Section 3.2 Correctness Check requirement)
    y_pred_all = np.clip(y_pred_all, 0, None)

    # 11. All Hours RMSE Evaluation
    rmse_all_hours = calculate_rmse(y_true_all, y_pred_all)
    print(f"Test-Set RMSE (All Hours): {rmse_all_hours:.2f} kW")

    # 12. Daytime Hours RMSE Evaluation (Strictly defined as IRRADIATION > 0)
    daytime_mask = test_df['irradiation'].values > 0 # Using renamed column
    y_true_daytime = y_true_all[daytime_mask]
    y_pred_daytime = y_pred_all[daytime_mask]

    if len(y_true_daytime) > 0:
        rmse_daytime = calculate_rmse(y_true_daytime, y_pred_daytime)
        print(f"Test-Set RMSE (Daytime Only - IRRADIATION > 0): {rmse_daytime:.2f} kW")
    else:
        print("No daytime rows found in the test set where IRRADIATION > 0.")

    # Plotting Actual vs. Predicted AC Power for all test hours
    plt.figure(figsize=(12, 6))
    sns.lineplot(x=test_df['datetime'], y=y_true_all, label='Actual AC Power', color='blue')
    sns.lineplot(x=test_df['datetime'], y=y_pred_all, label='Predicted AC Power', color='red', linestyle='--')
    plt.title('Actual vs. Predicted AC Power (Test Set - All Hours)')
    plt.xlabel('Date Time')
    plt.ylabel('AC Power (kW)')
    plt.legend()
    plt.grid(True)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

    # --- Placeholder for Training Metric ---
    # J_history_batch_gd = [...] # Your list of costs during training
    # if J_history_batch_gd:
    #     print(f"Cost J(theta) from training: {J_history_batch_gd[-1]:.4f}") # last recorded cost
    # else:
    #     print("Training cost history not available or implemented yet.")

except FileNotFoundError as e:
    print(f"Error loading file: {e}. Please ensure 'Plant_1_Generation_Data.csv' and 'Plant_1_Weather_Sensor_Data (1).csv' are in the /content/ directory.")
except KeyError as e:
    print(f"Error with column name: {e}. Please check the column names in your CSV files or the renamed columns.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
