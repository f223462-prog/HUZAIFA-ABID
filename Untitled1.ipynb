```python
import requests
import pandas as pd
import os

# Plant 1 coordinates (example for Kolkata, India, as implied by Asia/Kolkata timezone)
# You might need to adjust these if you have more precise coordinates for 'Plant 1'
LATITUDE = 22.57
LONGITUDE = 88.36

START_DATE = '2020-05-15'
END_DATE = '2020-06-17'
TIMEZONE = 'Asia/Kolkata'
HOURLY_VARIABLES = ['shortwave_radiation', 'temperature_2m', 'cloud_cover']

# Construct the API URL
base_url = 'https://archive-api.open-meteo.com/v1/archive'
params = {
    'latitude': LATITUDE,
    'longitude': LONGITUDE,
    'start_date': START_DATE,
    'end_date': END_DATE,
    'hourly': ','.join(HOURLY_VARIABLES),
    'timezone': TIMEZONE
}

# Make the API call
print(f"Calling Open-Meteo API for latitude={LATITUDE}, longitude={LONGITUDE}...")
response = requests.get(base_url, params=params)
response.raise_for_status() # Raise an exception for HTTP errors
data = response.json()
print("API call successful. Data received.")
```

```python
# Parse the JSON response into a DataFrame
hourly_data = data['hourly']

df_openmeteo = pd.DataFrame({
    'datetime': pd.to_datetime(hourly_data['time']),
    'sw radiation': hourly_data['shortwave_radiation'],
    'temp 2m': hourly_data['temperature_2m'],
    'cloud cover': hourly_data['cloud_cover']
})

# Localize Open-Meteo datetime to 'Asia/Kolkata' as the API data is already in this timezone
df_openmeteo['datetime'] = df_openmeteo['datetime'].dt.tz_localize('Asia/Kolkata')

# Display the first few rows of the DataFrame
display(df_openmeteo.head())

# Create the 'data' directory if it doesn't exist
output_dir = 'data'
os.makedirs(output_dir, exist_ok=True)

# Define the output file path
output_file_path = os.path.join(output_dir, 'plant1_openmeteo.csv')

# Save the DataFrame to a CSV file
df_openmeteo.to_csv(output_file_path, index=False)
print(f"Data successfully saved to {output_file_path}")
```

```python
sensor_file_path = 'sample_data/Plant_1_Weather_Sensor_Data (1).csv'
df_sensor = pd.read_csv(sensor_file_path)

# Rename 'DATE_TIME' to 'datetime'
df_sensor.rename(columns={'DATE_TIME': 'datetime'}, inplace=True)

# Convert to datetime objects and localize to 'Asia/Kolkata'
# Assume sensor data timestamps are also in 'Asia/Kolkata' local time
df_sensor['datetime'] = pd.to_datetime(df_sensor['datetime']).dt.tz_localize('Asia/Kolkata')

# Convert 'sw radiation' from W/m2 (Open-Meteo) to kW/m2 (sensor data)
df_openmeteo['sw radiation'] = df_openmeteo['sw radiation'] / 1000

# Merge the two DataFrames
df_merged = pd.merge(df_openmeteo, df_sensor, on='datetime', how='inner', suffixes=('_openmeteo', '_sensor'))

print(f"Sensor data loaded from {sensor_file_path}.")
print("Open-Meteo 'sw radiation' converted from W/m2 to kW/m2.")
display(df_merged.head())
```

```python
import matplotlib.pyplot as plt

def plot_and_compare_radiation(df, date):
    """Plots sensor and Open-Meteo radiation for a given date and reports peak hours."""
    df_day = df[df['datetime'].dt.date == pd.to_datetime(date).date()].copy()

    if df_day.empty:
        print(f"No data for {date}")
        return

    plt.figure(figsize=(12, 6))
    plt.plot(df_day['datetime'], df_day['IRRADIATION'], label='Sensor Irradiation (kW/m²)')
    plt.plot(df_day['datetime'], df_day['sw radiation'], label='Open-Meteo SW Radiation (kW/m²)')
    plt.xlabel('Time')
    plt.ylabel('Radiation (kW/m²)')
    plt.title(f'Radiation Comparison for {date}')
    plt.legend()
    plt.grid(True)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

    # Find peak hours
    peak_sensor_idx = df_day['IRRADIATION'].idxmax()
    peak_openmeteo_idx = df_day['sw radiation'].idxmax()

    peak_sensor_time = df_day.loc[peak_sensor_idx, 'datetime'].strftime('%H:%M')
    peak_openmeteo_time = df_day.loc[peak_openmeteo_idx, 'datetime'].strftime('%H:%M')

    print(f"--- Analysis for {date} ---")
    print(f"Sensor Peak: {peak_sensor_time} at {df_day.loc[peak_sensor_idx, 'IRRADIATION']:.2f} kW/m²")
    print(f"Open-Meteo Peak: {peak_openmeteo_time} at {df_day.loc[peak_openmeteo_idx, 'sw radiation']:.2f} kW/m²")

    # Check for shift
    time_diff_hours = abs((df_day.loc[peak_sensor_idx, 'datetime'] - df_day.loc[peak_openmeteo_idx, 'datetime']).total_seconds() / 3600)
    if time_diff_hours >= 1:
        print(f"Warning: Peak shift of {time_diff_hours:.2f} hours detected. Coordinates or time zone might be wrong.")
    else:
        print(f"Peak shift of {time_diff_hours:.2f} hours. Within acceptable range.")
    print("--------------------------")

# Select three days for verification
days_to_verify = ['2020-05-20', '2020-06-01', '2020-06-10']

for day in days_to_verify:
    plot_and_compare_radiation(df_merged, day)
```
